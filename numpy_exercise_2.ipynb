```python
# Importing numpy
import numpy as np
```

```python
# Creating an array with 12 values
arr = np.arange(0,12)
arr
```

```python
# Reshaping the array
re_arr = arr.reshape(3, 4)
re_arr 
```

```python
# Checking the shape of the new array
shape_arr = re_arr.shape
shape_arr
```

```python
# Select the first row
first_row_arr = re_arr[0]
first_row_arr
```

```python
# Select the last row
first_row_arr = re_arr[-1]
first_row_arr
```

```python
# Select the first column
first_row_arr = re_arr[:, 0]
first_row_arr
```

```python
# Select the last column
first_row_arr = re_arr[:, -1]
first_row_arr
```

```python
# Select the first two rows
first_row_arr = re_arr[:2]
first_row_arr
```

```python
# Select the last two columns
first_row_arr = re_arr[:, -2]
first_row_arr
```
