```python
import numpy as np
```

```python
# Create an array
num_array = np.arange(1, 17)
num_array
```

```python
# Create a 4x4 array
num_4x4_array = num_array.reshape(4, 4)
num_4x4_array
```

```python
# Filter values greater than 10
num_greater_than_10 = num_4x4_array[num_4x4_array > 10] 
print(num_greater_than_10)
```

```python
# Protect the dataset by creating a copy
original = np.arange(1, 17).reshape(4, 4)
original
```

```python
# Compare the original and modified copies of the dataset
modified = original.copy()
modified
```

# Explain why a copy can be safer
It preserves the original dataset and it is a good data practice.
